```python
#ax^2+bx+c=0
def pt2(a,b,c):
    delta=b**2-4*a*c
    if delta<0:
        return "Phuong trinh vo nghiem"
    elif delta==0:
        x=-b/(2*a)
        return f"Phuong trinh co nghiem duy nhat: {x}"
    elif delta>0:
        x1=(-b+math.sqrt(delta))/(2*a)
        x2=(-b-math.sqrt(delta))/(2*a)
        return f'Phuong trinh co 2 nghiem phan biet : {x1},{x2}'
try:
    a=float(input())
    b=float(input())
    c=float(input())
    ket_qua=pt2(a,b,c)
    print(ket_qua)
except ValueError:
    print("Vui long nhap so hop le")
```

```python
def doi_thoi_gian(x):
    gio = x // 3600
    phan_du = x % 3600
    phut = phan_du // 60
    giay = phan_du % 60
    return gio, phut, giay
if __name__ == "__main__":
    try:
        x = int(input("Nhập thời gian công việc (giây): "))
        if x < 0:
            print("Thời gian không thể là số âm!")
        else:
            gio, phut, giay = doi_thoi_gian(x)
            print(f"{gio} giờ, {phut} phút, {giay} giây")
    except ValueError:
        print("Vui lòng nhập vào một số nguyên hợp lệ")
```
