### Problem 4
#### Write a function that returns the count of positive integers k such that 1 <= k < n and k > (n - 1) / 2. (Example: n=9 -> 4).

```python
def fun(n):
    count=0
    for k in range(1,n):
        if k > (n-1)/2:
            count += 1
    return count
so = int(input())
fun(so)
```

### Problem 6
#### Write a function that computes the area of a triangle with side lengths a, b, c (use Heron's formula). Raise ValueError if the sides cannot form a triangle.

```python
import math
def tam_giac(a,b,c) :
    if not(a+b>c and a+c>b and b+c>a and a>0 and b>0 and c>0):
        raise ValueError("Các cạnh đã cho không thỏa mãn là cạnh của tam giác")
    p=(a+b+c)/2
    area=math.sqrt(p*(p-a)*(p-b)*(p-c))
    return area
s= tam_giac(3,4,5)
print(s)
```

### Problem 8
#### Write a function that returns the character that is n positions after character x in the alphabet.
#### Preserve case and wrap around within A..Z or a..z.

```python
def char(x,n):
    if 'a'<=x<= "z":
        return chr((ord(x) - ord('a') + n) % 26 + ord('a'))
    elif 'A'<=x<= "Z":
        return chr((ord(x) - ord('A') + n) % 26 + ord('A'))
print(char("Z",3))
```
