```python
import math
def la_so_nguyen_to(n):
    if n < 2:
        return False
    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True       
a,b=map(int,input().split())
tong = 0
for num in range(a, b + 1):
    if la_so_nguyen_to(num):
        tong += num
print(tong)
```

W4A8

```python
def dao_nguoc(n):
    dao=0
    while n>0:
        dao=dao*10+n%10
        n//=10
    return dao
n=int(input())
buoc=0
while n!=dao_nguoc(n):
    n=n+dao_nguoc(n)
    buoc+=1
print(buoc)
print(n)
```

```python
n=int(input())
ket_qua=[]
i=1
while i*i<=n:
    so_chinh_phuong=i*i
    chuoi_so=str(so_chinh_phuong)
    if len(chuoi_so)==len(set(chuoi_so)):
        ket_qua.append(chuoi_so)
    i+=1
if ket_qua:
    print(' '.join(ket_qua))
else:
    print("no number")
```
