```python
def is_armstrong(n:int)->bool:
    if n<0:
        return False
    am=0
    mu=int(len(str(n)))
    for i in str(n):
        am+=int(i)**mu
    return am==n
assert is_armstrong(153) == True
assert is_armstrong(9474) == True
assert is_armstrong(0) == True
assert is_armstrong(123) == False
assert is_armstrong(-153) == False
print("Pass all test")
```

```python
def sdx(n:int)->bool:
    if n<0:
        return False
    dao_nguoc=0
    goc=n
    while n>0:
        dao_nguoc=dao_nguoc*10 + n%10
        n//=10
    return dao_nguoc==goc
assert sdx(121) == True
assert sdx(12321) == True
assert sdx(0) == True
assert sdx(123) == False
assert sdx(-121) == False
print("Pass all test")
```

```python
import math
def detect_number_type(s: str) -> str:
    s = s.strip()
    try:
        int(s)
        return 'int'
    except ValueError:
        try:
            val = float(s)
            if math.isnan(val) or math.isinf(val):
                return 'neither'
            return 'float'
        except ValueError:
            return 'neither'
assert detect_number_type("123") == 'int'
assert detect_number_type("-45") == 'int'
assert detect_number_type("3.14") == 'float'
assert detect_number_type("-0.5") == 'float'
assert detect_number_type("abc") == 'neither'
assert detect_number_type("12.3.4") == 'neither'
assert detect_number_type("nan") == 'neither'
print("Pass all test")
```

```python
import math
def goc(x1, y1, x2, y2, x3, y3):
    tich_vo_huong = (x1 - x2) * (x3 - x2) + (y1 - y2) * (y3 - y2)
    do_dai = math.sqrt((x1 - x2)**2 + (y1 - y2)**2) * math.sqrt((x3 - x2)**2 + (y3 - y2)**2)
    gtct = max(-1.0, min(1.0, tich_vo_huong / do_dai))
    return round(math.degrees(math.acos(gtct)), 2)
assert goc(0, 1, 0, 0, 1, 0) == 90.0
assert goc(1, 0, 0, 0, 1, 1) == 45.0
assert goc(-1, 0, 0, 0, 1, 1) == 135.0
assert goc(0, 0, 1, 1, 2, 2) == 180.0
print("Pass all test")
```

```python
import math
def to_hop(n: int, k: int) -> int:
    if k < 0 or k > n:
        return 0
    return math.comb(n, k)
assert to_hop(5, 2) == 10
assert to_hop(5, 0) == 1
assert to_hop(5, 5) == 1
assert to_hop(4, 6) == 0
print("Pass all test")
```

```python
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    if car_type == 4:
        open_fare, rate_mid, rate_long = 12.0, 14.0, 11.0
    elif car_type == 7:
        open_fare, rate_mid, rate_long = 15.0, 16.0, 13.0
    else:
        raise ValueError("Chỉ hỗ trợ xe 4 chỗ hoặc 7 chỗ")
    if km <= 0:
        distance_fare = 0.0
    elif km <= 1:
        distance_fare = open_fare
    elif km <= 20:
        distance_fare = open_fare + (km - 1) * rate_mid
    else:
        distance_fare = open_fare + 19 * rate_mid + (km - 20) * rate_long
    wait_fare = max(0, wait_minutes - 5) * 0.5
    total = distance_fare + wait_fare
    if start_hour >= 22 or start_hour <= 5:
        total *= 1.1
    if km > 30:
        total *= 0.95
    return round(total, 2)

assert smart_taxi_fare(4, 10, 0.8, 3) == 12.0
assert smart_taxi_fare(4, 14, 10, 10) == 140.5
assert smart_taxi_fare(7, 23, 5, 0) == 86.9
assert smart_taxi_fare(4, 8, 35, 5) == 420.85
print("pass")
```

```python
def sum_1_to_n(n):
    if n <= 0:
        return 0
    tong = 0
    for x in range(1, n + 1):
        s = str(x)
        if len(s) == len(set(s)):
            tong += x
    return tong
assert sum_1_to_n(-5) == 0
assert sum_1_to_n(0) == 0
assert sum_1_to_n(10) == 55      
assert sum_1_to_n(22) == 220      
print("pass")
```
